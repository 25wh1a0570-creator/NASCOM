```python
def greet():
    print("Hello! Welcome to Python Functions.")

greet()
```

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Alice")
greet_user("Bob")
```

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Ravi")
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
number=[1,2,3,4,5,6]
#x=lambda a: a*a;
x=list(map(lambda a:a*a, number))
print(x)
y=list(filter(lambda a:a%2==0, number))
print(y)
```

```python

from functools import reduce
number=[1,2,3,4,5,6]
reduced=reduce(lambda x,y: x+y,number)
print(reduced)
```
